```python
def prosjecna_cijena(cijene):
    if not cijene:
        raise ValueError("Lista cijena ne smije biti prazna.")

    return sum(cijene) / len(cijene)
```

```python
cijene = [10, 20, 30, 40, 50]

rezultat = prosjecna_cijena(cijene)

print("Prosječna cijena:", rezultat)
```

```python
%%writefile cijene.py

def prosjecna_cijena(cijene):
    if not cijene:
        raise ValueError("Lista cijena ne smije biti prazna.")

    return sum(cijene) / len(cijene)
```

```python
%%writefile test_cijene.py

import pytest
from cijene import prosjecna_cijena


def test_prosjecna_cijena():
    assert prosjecna_cijena([10, 20, 30]) == 20


def test_jedna_cijena():
    assert prosjecna_cijena([15]) == 15


def test_prazna_lista():
    with pytest.raises(ValueError):
        prosjecna_cijena([])
```
